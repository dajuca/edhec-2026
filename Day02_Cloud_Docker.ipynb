### Why Docker in a data/BI course?

When you teach (or work) with **BigQuery + Python + APIs + Power BI**, the #1 pain is:  
> **“It works on my machine… but not on yours.”** 😅

Docker solves this by packaging **your app + your dependencies + your runtime** into a portable unit:

- ✅ same Python version for everyone  
- ✅ same libraries (pandas, google-cloud-bigquery, scikit-learn, etc.)  
- ✅ same environment variables (.env)  
- ✅ same command to run (no setup hell)

Think of Docker as a **standardized execution box** that runs the same way on Windows / macOS / Linux.

---

### Quick mental model

- **Image** 🧱 = a *recipe + snapshot* of a filesystem (Python + libs + your code)  
- **Container** 🚀 = a *running instance* of an image (your app running)  
- **Registry** 📦 = where images are stored (Docker Hub, GCP Artifact Registry, etc.)

```text
Dockerfile  -->  docker build  -->  Image  -->  docker run  -->  Container
```


### What Docker *really* does (under the hood)

Docker is not “a VM”. It uses OS-level isolation:

- **Namespaces**: isolate processes, network, filesystem views  
- **cgroups**: limit CPU/RAM usage  
- **Layered filesystem**: images are built in **layers** (fast rebuilds)

This is why containers are usually:
- ⚡ faster to start than VMs
- 📉 lighter than VMs
- ✅ perfect for reproducible teaching labs


### The 3 files you’ll use all the time

1. **`Dockerfile`**: how to build the image (runtime + dependencies + app)
2. **`docker-compose.yml`**: how to run multiple services together (app + db + admin UI + etc.)
3. **`.env`**: secrets / configuration (never commit real secrets)

> **Rule of thumb**:  
> - `Dockerfile` = *how to build*  
> - `docker-compose.yml` = *how to run*  
> - `.env` = *how to configure*


## 🧱 1) The Dockerfile (what to put inside)

A Dockerfile is a **step-by-step build recipe**.  
For Python data apps, your Dockerfile should usually contain:

✅ **Base image** (pin versions!)  
✅ **System dependencies** (optional)  
✅ **Python dependencies** (`pip install -r requirements.txt`)  
✅ **Copy your source code**  
✅ **Default command** (`CMD`)

---

### A clean Dockerfile for a BigQuery + ML project

```dockerfile
# =========================
# 1) Choose a base image
# =========================
FROM python:3.12-slim

# =========================
# 2) Basic container hygiene
# =========================
ENV PYTHONDONTWRITEBYTECODE=1 \
    PYTHONUNBUFFERED=1

WORKDIR /app

# (Optional) if you need build tools (often not needed)
# RUN apt-get update && apt-get install -y --no-install-recommends \
#     build-essential && rm -rf /var/lib/apt/lists/*

# =========================
# 3) Install dependencies
# =========================
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# =========================
# 4) Copy code
# =========================
COPY . .

# =========================
# 5) Default command
# =========================
# Example: run a module that fetches data from BQ and trains a model
CMD ["python", "-m", "luxury.interface.main"]
```

### What belongs in `requirements.txt`?

For a BigQuery + ML mini-project, a minimal example:

```txt
pandas
numpy
scikit-learn
pyarrow
google-cloud-bigquery
google-cloud-bigquery-storage
db-dtypes
python-dotenv
```


## 🧩 2) Docker Compose (what to put inside)

`docker-compose.yml` is for **running multiple services** together.

Typical BI/data stack examples:
- `app` (Python: ETL/ML/API)
- `postgres` (optional if you need a DB)
- `adminer`/`pgadmin` (optional UI)

Even if you only have one service, Compose is useful because it standardizes:
- ports
- volumes
- environment variables
- dependencies
- networks

---

### A Compose file for a Python service that talks to BigQuery

```yaml
services:
  app:
    build: .
    container_name: luxury_app
    env_file:
      - .env
    volumes:
      - .:/app:cached
      # 🔐 Mount your service account key (JSON) into the container (read-only)
      - ./secrets/gcp-sa.json:/secrets/gcp-sa.json:ro
    environment:
      # Tell Google SDK / BigQuery client where to find the key
      - GOOGLE_APPLICATION_CREDENTIALS=/secrets/gcp-sa.json
    command: ["python", "-m", "luxury.interface.main"]
```

### What **should** go in compose?

✅ `env_file` + `environment`  
✅ `volumes` (for live coding + credentials mounting)  
✅ `command` (what runs)  
✅ `ports` (if you expose an API / dashboard)

### What should **NOT** go in compose?

❌ hard-coded secrets (tokens, private keys)  
❌ huge logic (keep logic in your Python code)


## 🔐 3) BigQuery credentials inside Docker (service account key)

When your container runs, it has **its own filesystem**.  
So your BigQuery code cannot magically “see” your local credentials unless you:

- mount them into the container (recommended for teaching), or
- use workload identity / metadata server (production on GCP)

For this course/workshop we use the simplest and most explicit method:

✅ **Service account key (JSON) mounted into the container**

Put it in yout secrets folder (that should be gitignored !)
```text
your_project/
  secrets/
    gcp-sa.json        <-- DO NOT COMMIT
  .gitignore
  docker-compose.yml
  Dockerfile
  ...
```

5) Add this to `.gitignore`:

```gitignore
# Secrets
secrets/
*.json
.env
```

6) In `docker-compose.yml` mount the file and set:

```yaml
- GOOGLE_APPLICATION_CREDENTIALS=/secrets/gcp-sa.json
```

That’s it. Any code using `google-cloud-bigquery` will authenticate automatically.

# 🐳 Docker Repository Structure (Very Important)

When we create a **Docker-based data project**, we are not just creating a container.
We are designing a **mini data platform**.

Each file has **one responsibility**.



## 🧱 Dockerfile – Build ONE reusable image

The **Dockerfile** answers one question:

> How do we build a reproducible execution environment?

Key idea:
- ONE Dockerfile
- ONE image
- MANY services

This image contains:
- Python
- Libraries
- Your code

But **no secrets** and **no business logic choice**.



## 🎼 Docker Compose – Orchestrating multiple services

Docker Compose answers a different question:

> How do we run **multiple containers together**, each with a different role?

Compose introduces the idea of **services**.



## 🛠️ Lifecycle commands students must know

```bash
docker compose up --build
docker compose up api
docker compose logs cleaner
docker compose down
```

These commands control the **entire platform**.
