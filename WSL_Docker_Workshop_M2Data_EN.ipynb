# 1) 🪟 Installing WSL2 on Windows (step by step)

## 🐧 What is WSL?
**WSL (Windows Subsystem for Linux)** allows you to run a Linux environment **directly inside Windows**, without dual-boot or heavy virtual machines.

### WSL1 vs WSL2 (simple explanation)
- **WSL1**: system-call translation, limited compatibility
- **WSL2**: real Linux kernel, much better compatibility and performance

👉 For Data & Docker, we always use **WSL2**.

---

## 🧩 Architecture overview

![WSL1 vs WSL2 architecture](https://4.bp.blogspot.com/-e_FaEsP7nH8/Xca4QMFtjQI/AAAAAAAAERA/r59YN5UCXlwdNsfPNy-sjjyTQ4o_dtbjACLcBGAsYHQ/s1600/wsl1-wsl2.png)


## ✅ Prerequisites
- Windows 10 or 11 (Windows 11 recommended)
- Administrator rights
- Hardware virtualization enabled (VT-x / AMD-V)

---

## 1.1 Open PowerShell **as Administrator** 👮‍♂️
1. Start Menu → type **PowerShell**
2. Right-click → **Run as administrator**


## 1.2 Install WSL (one-command method) ⚡

In PowerShell (Admin):

```powershell
wsl --install
```

✅ This installs:
- WSL2
- A default Linux distribution (usually Ubuntu)

🔁 A reboot may be required.


## 1.3 Verify WSL installation 🧪

```powershell
wsl -l -v
```

You should see something like:
- `Ubuntu   Running   2`


## 1.4 Force WSL2 if needed (very common)

```powershell
wsl --set-default-version 2
wsl --set-version Ubuntu 2
```

> Replace `Ubuntu` with the exact distro name from `wsl -l -v`.


## 1.5 First Ubuntu launch 🥳
When opening Ubuntu for the first time:
- choose a **username**
- choose a **password**
  - 🔒 nothing appears when typing the password — this is normal


## 1.6 Update Ubuntu packages 🧼

Inside Ubuntu:

```bash
sudo apt update
sudo apt upgrade -y
```


## 1.7 Where should you store your projects? (IMPORTANT) ⚡

You have two worlds in WSL:
- 🪟 Windows filesystem: `/mnt/c/...`
- 🐧 Linux filesystem: `~/...` (**recommended**)

Best practice for Docker & Data tools:
```bash
cd ~
mkdir -p projects
cd projects
```


## 1.8 Quick Linux sanity check ✅

```bash
uname -a
python3 --version
```

# 3) 🔬 Mini Lab (10–15 min): First Docker container

🎯 Goal: Build and run a simple Python container.

## Steps
1. Create folder `hello-docker/`
2. Create `app.py`
3. Create `Dockerfile`
4. Build & run

---

## app.py
```python
print("👋 Hello from Docker! 🐳")
```

## Dockerfile
```dockerfile
FROM python:3.11-slim

WORKDIR /app
COPY app.py /app/app.py

CMD ["python", "app.py"]
```

## Build & Run
```bash
docker build -t hello-data .
docker run --rm hello-data
```

# 4) 🧯 Troubleshooting

- WSL2 engine not enabled
- WSL integration not enabled
